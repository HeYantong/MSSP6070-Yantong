```python
# a. A function only executes when it is called. When you only define it, it just knows what should do but doesn't run yet.

# b. There is only one global scope. But there can be many local scopes at the same time. A local scope can be directly created whenever a function is called.

# c. Inside the function, the local scope exists. However, when the function call returns, the local scope destroys.

# d. Use try and except statements to prevent crashing when it meets an error. You can see an example in my code below.
```

```python
def collatz(number):
  if number % 2 == 0:
    return (number // 2)
  else:
    return (number * 3 + 1)

try:
    number = int(input("Enter a number:\n"))
except ValueError:
    print("Please enter an integer.")

while number != 1:
    number = collatz(number)
    print(number, end=' ')
```
