
In the context of downloading data from Yahoo Finance using the yfinance library, the period parameter defines the range of data to download and can have the following values:

*    1d, 5d: data for the past 1 or 5 days

*   1mo, 3mo, 6mo: data for the past 1, 3, or 6 months
*   1y, 2y, 5y, 10y, ytd: data for the past 1, 2, 5, or 10 years, or year to date.


*   max: data for the maximum available period.


These values allow you to specify the time frame for which you want to retrieve historical data for a given stock.

#Task 1: Read into seperate data frame the following tickers: Apple, Facebook, Google, and Netflix; use Yahoo finance.

```python
Apple = yf.Ticker("AAPL")
Facebook = yf.Ticker("META")
Google = yf.Ticker("GOOGL")
Netflix = yf.Ticker("NFLX")

startDate = "2023-01-01"
endDate = "2024-01-01"

apple_df = Apple.history(start=startDate, end=endDate)
facebook_df = Facebook.history(start=startDate, end=endDate)
google_df = Google.history(start=startDate, end=endDate)
netflix_df = Netflix.history(start=startDate, end=endDate)

display(apple_df.head())
display(facebook_df.head())
display(google_df.head())
display(netflix_df.head())
```

#Task 2:  Join the individual stock data frames into a single data frame.

```python
import pandas as pd

apple_df["Ticker"] = "AAPL"
facebook_df["Ticker"] = "META"
google_df["Ticker"] = "GOOGL"
netflix_df["Ticker"] = "NFLX"

combined_df = pd.concat([
    apple_df,
    facebook_df,
    google_df,
    netflix_df
])

combined_df = combined_df.reset_index()

combined_df.head()
```

# Task 3: Create a visualization of the following financial metrics for each stock:  quickRatio, currentRatio, totalRevenue, debtToEquity, revenuePerShare, returnOnAssets, returnOnEquity, grossProfits


```python
apple_info = Apple.info
facebook_info = Facebook.info
google_info = Google.info
netflix_info = Netflix.info
```

```python
metrics = [
    "quickRatio",
    "currentRatio",
    "totalRevenue",
    "debtToEquity",
    "revenuePerShare",
    "returnOnAssets",
    "returnOnEquity",
    "grossProfits"
]

financial_df = pd.DataFrame({
    "AAPL": [apple_info.get(m) for m in metrics],
    "META": [facebook_info.get(m) for m in metrics],
    "GOOGL": [google_info.get(m) for m in metrics],
    "NFLX": [netflix_info.get(m) for m in metrics]
}, index=metrics)

financial_df = financial_df.T
financial_df = financial_df.round(2)

financial_df
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(4, 2, figsize=(14, 16))

colors = ["#8DA0CB", "#66C2A5", "#FC8D62", "#E78AC3"]

for i, metric in enumerate(metrics):
    ax = axes.flatten()[i]

    financial_df[metric].plot(
        kind="bar",
        ax=ax,
        color=colors,
        rot=0
    )

    ax.set_title(metric)
    ax.set_xlabel("")
    ax.grid(axis="y", alpha=0.2)
```

#Task 4: Create a single boxplot for the Closing price for each stock.


```python
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=combined_df,
    x="Ticker",
    y="Close",
    hue="Ticker",
    palette="Set2",
    width=0.6,
    legend=False
)

plt.title("Distribution of Stock Closing Prices in 2023")
plt.xlabel("Stock Ticker")
plt.ylabel("Closing Price (USD)")
plt.grid(axis="y", alpha=0.2)

plt.show()
```

# Task 5: Create a daily trading volume visualization for the year 2023 for each stock.


```python
plt.figure(figsize=(10, 5))

sns.lineplot(
    data=combined_df,
    x="Date",
    y="Volume",
    hue="Ticker",
    palette="Set2",
    linewidth=1
)

plt.title("Daily Trading Volume by Stock in 2023")
plt.xlabel("Date")
plt.ylabel("Trading Volume")
plt.legend(title="Stock Ticker")
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```
