```python
import pandas as pd

url = (
    "https://raw.githubusercontent.com/"
    "HeYantong/MSSP6070-Yantong/main/WeeklyModules/Week04/parsed.csv"
)

df = pd.read_csv(url)
```

```python
japan_mb = df.loc[
    (df["parsed_place"] == "Japan") &
    (df["magType"] == "mb"),
    "mag"
]

float(japan_mb.quantile(0.95))
```

```python
indonesia = df.loc[
    (df["parsed_place"] == "Indonesia")
]

tsunami_count = indonesia.loc[
    indonesia["tsunami"] == 1,
    "tsunami"
].count()

total_count = indonesia["tsunami"].count()

percentage = tsunami_count / total_count * 100

print(f"{percentage:.2f}%")
```

```python
nevada = df.loc[
    df["parsed_place"] == "Nevada"
]

nevada.describe()

# The standard deviation for mmi is NaN because there is only one non-missing mmi observation for earthquakes in Nevada.
```

```python
ring_of_fire_locations = [
    "Bolivia",
    "Chile",
    "Ecuador",
    "Peru",
    "Costa Rica",
    "Guatemala",
    "Mexico",
    "Japan",
    "Philippines",
    "Indonesia",
    "New Zealand",
    "Canada",
    "Fiji",
    "Alaska",
    "Washington",
    "California",
    "Russia",
    "Taiwan",
    "Tonga",
    "Kermadec Islands"
]

df["ring_of_fire"] = (
    df["parsed_place"].isin(ring_of_fire_locations)
    |
    df["parsed_place"].str.contains(
        "Antarctic",
        case=False,
        na=False
    )
)

print(
    df[["parsed_place", "ring_of_fire"]]
)
```

```python
ring_of_fire_count = (
    df["ring_of_fire"] == True
).sum()

non_ring_of_fire_count = (
    df["ring_of_fire"] == False
).sum()

print("Ring of Fire earthquakes:",ring_of_fire_count)
print("Outside earthquakes:",non_ring_of_fire_count)
```

```python
ring_of_fire_tsunami_count = df.loc[
    (df["ring_of_fire"] == True) &
    (df["tsunami"] == 1),
    "tsunami"
].count()

print(
    "Tsunami count along the Ring of Fire:",
    ring_of_fire_tsunami_count
)
```
