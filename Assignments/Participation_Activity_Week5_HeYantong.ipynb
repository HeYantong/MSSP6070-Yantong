```python
import pandas as pd

fb = pd.read_csv("data/fb.csv")
aapl = pd.read_csv("data/aapl.csv")
amzn = pd.read_csv("data/amzn.csv")
nflx = pd.read_csv("data/nflx.csv")
goog = pd.read_csv("data/goog.csv")

aapl.head()
```

```python
fb["ticker"] = "FB"
aapl["ticker"] = "AAPL"
amzn["ticker"] = "AMZN"
nflx["ticker"] = "NFLX"
goog["ticker"] = "GOOG"

aapl.head()
```

```python
faang = pd.concat([fb, aapl, amzn, nflx, goog], ignore_index=True)
```

```python
faang.to_csv("data/faang.csv", index=False)
```

```python
faang["date"] = pd.to_datetime(faang["date"])
faang["volume"] = faang["volume"].astype(int)
faang = faang.sort_values(["date", "ticker"])

faang.nlargest(7, "volume")
```

```python
melted = faang.melt(
    id_vars=["date", "ticker"]
)

melted.head()
```
